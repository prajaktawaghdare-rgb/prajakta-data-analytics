## Numpy copy vs view

```python
# Create a NumPy array x1
x1 = np.array([1, 2, 3, 4, 5])
```

```python
# Assign x2 = x1 (both names refer to the same array)
x2 = x1
```

```python
# Print x1 and x2
print('x1:', x1)
print('x2:', x2)
```

Ok now you have seen that both of them are same

```python
# Change the first element of x2 to 10
x2[0] = 10
```

```python
# Again print x1 and x2
print('x1:', x1)
print('x2:', x2)
```

```python
# Create a view of x1 and store it in x3
x3 = x1.view()
```

```python
# Check whether x1 and x3 share memory
print(np.shares_memory(x1, x3))
```

Woh! simple assignment is similar to view. That means
The view does not own the data and any changes made to the view will affect the original array, and any changes made to the original array will affect the view.

Don't agree? ok lets change x3 and see if original array i.e. x1 also changes

```python
# Change the first element of x3 to 100
x3[0] = 100
```

```python
# Print x1 and x3 to check whether changes are reflected in both
print('x1:', x1)
print('x3:', x3)
```

Now its proved.

Lets see how **Copy()** function works

```python
# Create an array x4 which is a copy of x1
x4 = x1.copy()
```

```python
# Change the last element of x4 to 900
x4[-1] = 900
```

```python
# Print x1 and x4 to check whether changes are reflected in both
print('x1:', x1)
print('x4:', x4)
```

Hey! such an intresting output. You noticed buddy! your original array didn't get changed on change of its copy ie. x4.

Still not convinced? Ok lets see if they both share memory or not

```python
# Check whether x1 and x4 share memory
print(np.shares_memory(x1, x4))
```

**hstack vs vstack function**

Stacking is similar to concatenation, but stacking functions have their own axis behavior.

1. `hstack()` joins arrays horizontally.
2. `vstack()` stacks arrays vertically (as rows).


```python
# Stack x1 and x4 horizontally
np.hstack((x1, x4))
```

```python
# Stack x1 and x4 vertically
np.vstack((x1, x4))
```

We hope now you saw the difference between them.

Fun fact! you can even use concatenate() function to join 2 arrays along with the axis. If axis is not explicitly passed, it is taken as 0 ie. along column

Lets try this function as well

```python
# Define two 2-D arrays and join along rows using concatenate()
arr1 = np.array([[1, 2, 3], [4, 5, 6]])
arr2 = np.array([[7, 8, 9], [10, 11, 12]])
np.concatenate((arr1, arr2), axis=0)
```

```python
# Join arr1 and arr2 along columns using concatenate()
np.concatenate((arr1, arr2), axis=1)
```
